```python
def env_state():
  states=["S0","S1"]
  return states
```

```python
states=env_state()
states
```

```python
def env_action():
  actions=["up","down","left","right"]
  return actions
```

```python
actions=env_action()
actions
```

```python
def env_df():
  gama=0.9
  return gama
```

```python
gamma=env_df()
gamma
```

```python
def env_rewards():
  rewards={
      "Move":-1,
      "Wall":-1,
      "Goal":10
  }
  return rewards
```

```python
rewards=env_rewards()
rewards
```

```python
def env_transition():
  transition={
      "S0":{"up":"S0","down":"S1","left":"S0","right":"S0"},
      "S1":{"up":"S0","down":"S1","left":"S1","right":"Goal"}
  }
  return transition
```

```python
transitions=env_transition()
transitions
```

```python
next_state=transitions["S1"]["right"]
next_state
```

```python
def rewards_cal(Next_state):
  if Next_state=="Goal":
    reward=rewards["Goal"]
  elif Next_state=="Wall":
    reward=rewards["Wall"]
  else:
    reward=rewards["Move"]
  return reward
```

```python
state="S1"
action="up"
next_state=transitions[state][action]
print(f"Next State: {next_state}")
reward=rewards_cal(next_state)
print(f"Reward: {reward}")
```

```python
def env_policy():
  policy={
      "S0":"up",
      "S1":"up"
  }
  return policy
```

```python
# Initial arbitrary policy
policy = env_policy()

# Initial state values
V = {
    "S0": 0.0,
    "S1": 0.0,
    "Goal": 0.0
}

theta = 1e-8
max_iterations = 10000

print("\nInitial Policy:", policy)
```

```python
def policy_evaluation(policy, V):

    for iteration in range(max_iterations):

        old_V = V.copy()
        new_V = old_V.copy()

        for state in states:

            # Action selected by the current policy
            action = policy[state]

            # Deterministic next state
            next_state = transitions[state][action]

            # Immediate reward
            r = rewards_cal(next_state)

            # Bellman expectation update for a fixed policy
            new_V[state] = r + gamma * old_V[next_state]

        # Check convergence
        delta = max(
            abs(new_V[s] - old_V[s])
            for s in states
        )

        V = new_V

        if delta < theta:
            break

    return V
```

```python
V = policy_evaluation(policy, V)
V
```

```python
def calculate_q(state, action, V):

    next_state = transitions[state][action]
    r = rewards_cal(next_state)

    return r + gamma * V[next_state]
```

```python
calculate_q("S0", "up", V)
```

```python
while True:

    # Policy Evaluation
    V = policy_evaluation(policy, V)

    # Policy Improvement
    improved_policy = {}

    for state in states:
        q_values = {
            action: calculate_q(state, action, V)
            for action in actions
        }

        improved_policy[state] = max(
            q_values, key=q_values.get
        )

    # Check policy stability
    if improved_policy == policy:
        break

    policy = improved_policy

print("Optimal State Values:", V)
print("Optimal Policy:", policy)
```
